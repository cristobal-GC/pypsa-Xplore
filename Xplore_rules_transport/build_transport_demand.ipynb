# Rule: **build_transport_demand**

**Description**

Build land transport demand per clustered model region including efficiency
improvements due to drivetrain changes, time series for electric vehicle
availability and demand-side management constraints.

**Outputs**

- resources/`transport_demand_s_{clusters}.csv`
- resources/`transport_data_s_{clusters}.csv`
- resources/`avail_profile_s_{clusters}.csv`
- resources/`dsm_profile_s_{clusters}.csv`

```python
######################################## Parameters

### Run
name = 'case_SectorCoupled'
prefix = ''

### Network
clusters = 5
```

```python
##### Import packages
import matplotlib.pyplot as plt
import os 
import sys
import pandas as pd


##### Import local functions
sys.path.append(os.path.abspath(os.path.join('..')))
import functions as xp


##### Read params.yaml
params = xp.read_params('../params.yaml')


##### Ignore warnings
import warnings
warnings.filterwarnings('ignore', category=UserWarning)
```

## `transport_demand_s_{clusters}.csv`

Load the file and preview its content.

Units: 100 vehicle-km.

```python
file = f"transport_demand_s_{clusters}.csv"

df_transport_demand = xp.load_file_csv(
    params,
    file,
    prefix=prefix,
    name=name,
    location="resources",
    index_col=0,
)

print(f'Total mill. vehicle-km: {100*df_transport_demand.sum().sum()/1e6:.2f}')

df_transport_demand.head()

```

Represent as a time series.

```python
#################### Parameters
start = '2013-01-01'
end = '2013-01-07'



#################### Figure

# Try datetime parsing for a cleaner x-axis when snapshots are timestamps.
df_transport_demand.index = pd.to_datetime(df_transport_demand.index, errors="ignore")

fig, ax = plt.subplots(figsize=(12, 4))
df_transport_demand.loc[start:end].plot(ax=ax, legend=True, alpha=0.8, linewidth=1.0, marker='o', markersize=4)
ax.set_title("Transport demand by region")
ax.set_xlabel("snapshot")
ax.set_ylabel("demand [100 vehicle-km]")
ax.grid(True, color='lightgray', alpha=0.5)
plt.tight_layout()
plt.show()
```

## `avail_profile_s_{clusters}.csv`

Profile of plugged-in availability for passenger electric vehicles per hour and clustered region.

Load the file and preview its content.

```python
file = f"avail_profile_s_{clusters}.csv"

df_avail_profile = xp.load_file_csv(
    params,
    file,
    prefix=prefix,
    name=name,
    location="resources",
    index_col=0,
)

df_avail_profile.head()
```

Represent as a time series.

```python
#################### Parameters
start = '2013-01-01'
end = '2013-01-07'



#################### Figure


# Try datetime parsing for a cleaner x-axis when snapshots are timestamps.
df_avail_profile.index = pd.to_datetime(df_avail_profile.index, errors="ignore")

fig, ax = plt.subplots(figsize=(12, 4))
df_avail_profile.loc[start:end].plot(ax=ax, legend=False, alpha=0.8, linewidth=1.0, marker='o', markersize=4)
ax.set_title("EV availability profile by region")
ax.set_xlabel("snapshot")
ax.set_ylabel("availability [p.u.]")
ax.grid(True, color='lightgray', alpha=0.5)
plt.tight_layout()
plt.show()

df_avail_profile.head()
```

## `dsm_profile_s_{clusters}.csv`

Profile for DSM: SOC required per hour and clustered region.

Load the file and preview its content.

```python
file = f"dsm_profile_s_{clusters}.csv"

df_dsm_profile = xp.load_file_csv(
    params,
    file,
    prefix=prefix,
    name=name,
    location="resources",
    index_col=0,
)

df_dsm_profile.head()
```

Represent as a time series.

```python
#################### Parameters
start = '2013-01-01'
end = '2013-01-01'



# Try datetime parsing for a cleaner x-axis when snapshots are timestamps.
df_dsm_profile.index = pd.to_datetime(df_dsm_profile.index, errors="ignore")

fig, ax = plt.subplots(figsize=(12, 4))
df_dsm_profile.loc[start:end].plot(ax=ax, kind='bar', legend=False, alpha=0.8)
ax.set_title("DSM profile by region")
ax.set_xlabel("snapshot")
ax.set_ylabel("availability [p.u.]")
ax.tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.show()

df_dsm_profile.head()
```
