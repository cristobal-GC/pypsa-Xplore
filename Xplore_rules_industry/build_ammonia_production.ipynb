# Rule: **build_ammonia_production**

**Description**

This rule calculates ammonia (NH₃) production by converting values reported by the USGS in terms of nitrogen content into ammonia units using the corresponding molecular weight ratio. In addition, the selected country is identified and harmonised to standard country codes to ensure compatibility with the network.

**Outputs**

- resources/{prefix}/{name}/`ammonia_production.csv`

```python
######################################## Parameters

### Run
prefix = ''
name = ''

### Spatial domain 'ES' or 'EU' (for maps domain and NUTS regions)
spatial_domain = 'ES'
country_name = 'Spain'
```

```python
##### Imports
import sys
import os

##### Import local functions
sys.path.append(os.path.abspath(os.path.join('..')))
import functions as xp

##### Read params.yaml
params = xp.read_params('../params.yaml')

##### Ignore warnings
import warnings
warnings.filterwarnings('ignore', category=UserWarning)
```

## `ammonia_production.csv`  
Load the file and preview its content.

```python
file = f"ammonia_production.csv"

ammonia_prod = xp.load_file_csv(
    params,
    file,
    prefix=prefix,
    name=name,
    location="resources",
)

ammonia_prod = ammonia_prod.set_index(ammonia_prod.columns[0])

ammonia_prod.T
```

See the ammonia production for Spain.

```python
ammonia_ES = ammonia_prod.loc[spatial_domain]
print("Calculated ammonia production for Spain (kt NH₃/year) according to USGS ammonia production reports:")
ammonia_ES
```

See the ammonia production for any other country on the list.

```python
# Specify a country code from the available countries
country = "BR"
ammonia_country = ammonia_prod.loc[country]
print(f"Calculated ammonia production for {country} (kt NH₃/year) according to USGS ammonia production reports:")
ammonia_country
```
