# Rule: **build_ammonia_production**

**Description**

This rule retrieves raw data on historical annual ammonia (NH₃) production by country, reported in terms of its nitrogen content, from the United States Geological Survey (USGS) statistics. Data available for 2018–2022.

**Outputs**

- data/nitrogen_statistics/archive/2022/`nitro-ert.xlsx`

```python
######################################## Parameters

### Run
prefix = ''
name = ''

### Spatial domain 'ES' or 'EU' (for maps domain and NUTS regions)
spatial_domain = 'ES'
country_name = 'Spain'
```

```python
##### Imports
import pandas as pd
from pathlib import Path
import sys
import os

##### Import local functions
sys.path.append(os.path.abspath(os.path.join('..')))
import functions as xp

##### Read params.yaml
params = xp.read_params('../params.yaml')

##### Ignore warnings
import warnings
warnings.filterwarnings('ignore', category=UserWarning)

##### Load the file
nitrogen_path = (Path(params["rootpath"]) / "data" / "nitrogen_statistics" / "archive"/ "2022"/ f"nitro-ert.xlsx")

##### Set options
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", None)
```

## `nitro-ert.xlsx`  
Load the file and preview its content.

```python
usgs_raw = pd.read_excel(
    nitrogen_path,
    sheet_name="T12",
    skiprows=5,
    header=0,
    index_col=0,
    skipfooter=7,
    na_values=["--"],
)

usgs_raw.columns = usgs_raw.columns.astype(str)
usgs_clean = usgs_raw.loc[
    :, usgs_raw.columns.str.fullmatch(r"\d{4}")
]

years = ["2018", "2019", "2020", "2021", "2022"]
usgs_clean = usgs_clean[years]

usgs_clean.T
```

See the reported values for a specific country

```python
# Choose a country from the options available in the USGS report
# # Note: some countries have an extra character at the end, write the same word as in the report
country = "Spain" 

usgs_country = usgs_clean.loc[country]

print(f"Ammonia production for {country} (kt contained nitrogen/year):")
usgs_country
```
