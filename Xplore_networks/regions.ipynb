# Xplore: **regions**

Onshore and offshore regions of a PyPSA-$X$ network. They change only at these stages
of the workflow (`compose_network` and `solve_network` use the clustered regions):

`base_network` · `simplify_network` · `cluster_network`

```python
######################################## Parameters

### Rule whose regions you want to analyse (uncomment one of the following):
# rule = 'base_network'          # resources/onshore_regions_base.geojson
# rule = 'simplify_network'      # resources/onshore_regions_simplified.geojson
rule = 'cluster_network'         # resources/onshore_regions.geojson

### Run (folder resources/<prefix>/<name>/)
name = ''
prefix = ''
```

```python
##### Import packages
import os
import sys

##### Import local functions
sys.path.append(os.path.abspath(os.path.join("..")))
import functions as xp

##### Read params.yaml
params = xp.read_params("../params.yaml")
```

```python
##### Region files (onshore / offshore) matching this rule
src = xp.resolve_network_source(rule)
print(f"regions : {src['regions_onshore']} / {src['regions_offshore']}")

gdf_regions_onshore, gdf_regions_offshore = xp.load_regions(
    params,
    prefix=prefix,
    name=name,
    filenames=(src['regions_onshore'], src['regions_offshore']),
)
```

## `regions_onshore`

```python
display(gdf_regions_onshore.head())
print(f"Number of onshore regions: {len(gdf_regions_onshore)}")
gdf_regions_onshore.plot()
```

## `regions_offshore`

```python
display(gdf_regions_offshore.head())
print(f"Number of offshore regions: {len(gdf_regions_offshore)}")
gdf_regions_offshore.plot()
```
